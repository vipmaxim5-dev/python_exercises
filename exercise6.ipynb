```python
from datetime import datetime # импорт datetime библиотеки

""" Функция получения числового значения от пользователя """
def input_int(string):
    while True:
        user_input = input(string) # Получаем строку от пользователя
        
        # Проверяем, состоит ли строка только из цифр
        if user_input.isdigit():
            integer = int(user_input) # Переводим в число
            if integer > 24:
                print('Ты вводишь какую то чушь! Еще раз')
            else:
                return integer # Выходим из цикла и возвращаем число
        elif user_input == '':
            print('Пользователь ничего не ввел, но я же умный, я посмотрю какое время у него на ПК')
            integer = datetime.now().hour # Получаем текущий час с системного времяни\
            return integer
        else: # Если вводиться не число
            print('Еще раз!')

hour = input_int('Который сейчас час?:')

if 0 <= hour < 6: day_time = "Ночь"
elif 6 <= hour < 12: day_time = "Утро"
elif 12 <= hour < 18: day_time = "День"
else: day_time = "Вечер"

print(f'Сейчас {day_time}')
```
