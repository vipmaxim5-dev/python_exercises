```python
usd_rate, eur_rate,cny_rate = 90.5, 98.2, 12.3

""" Функция получения числового значения от пользователя """
def input_int(string):
    while True:
        user_input = input(string) # Получаем строку от пользователя
        
        # Проверяем, состоит ли строка только из цифр
        if user_input.isdigit():
            integer = int(user_input) # Переводим в число
            return integer # Выходим из цикла и возвращаем число
        else:
            print('Неверно, введите еще раз')

""" Функция расчета валют """
def get_rate(rubles, rate):
    rate_result = rubles / rate # делим рубли на валютную ставку
    rate_result = f"{rate_result:.2f}" # округляем до 2х знаков после запятой
    return rate_result # возвращаем готовый результат

rubles = input_int('Введите сумму в рублях (например, 5000):')

usd_rate_result = get_rate(rubles, usd_rate)
eur_rate_result = get_rate(rubles, eur_rate)
cny_rate_result = get_rate(rubles, cny_rate)

print('На эти деньги можно купить:')
print(f'{usd_rate_result} доллар(а/ов)')
print(f'или {eur_rate_result} евро')
print(f'или {cny_rate_result} юан(ь/ий)')
```
