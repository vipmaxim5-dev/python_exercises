```python
phonebook = {
    "Иванов": "123-45-67",
    "Петров": "234-56-78",
    "Сидорова": "345-67-89"
}

phonebook["Смирнов"] = "456-78-90"
print(f'Добавлен контакт Смирнов: {phonebook}')

phonebook["Петров"] = "999-99-99"
print(f'Изменен телефон Петрова: {phonebook}')

del phonebook["Сидорова"]
print(f"Удален контакт Сидоровой:", phonebook)

is_ivanov_present = "Иванов" in phonebook
print(f"Есть ли в справочнике контакт 'Иванов'?: {is_ivanov_present}")

print(f'Список всех контактов:')
for name, phone in phonebook.items():
    print(f'{name}: {phone}')
    
search_phone = "234-56-78"
found_name = None

for name, phone in phonebook.items():
    if phone == search_phone:
        found_name = name
        break

if found_name:
    print(f"По номеру {search_phone} найден контакт: {found_name}")
else:
    print(f"Контакт с номером {search_phone} не найден")
```
