```python
""" Функция получения строку состоящую из числе от пользователя """
def input_str(string):
    while True:
        user_input = input(string) # Получаем строку от пользователя
        if len(user_input) == 6 and user_input.isdigit():
            return user_input # Возвращаем СТРОКУ
        else:
            print('Неверно, введите еще раз')
            
ticket_number = input_str('Введите шестизначное число (например, 123456):')

# Разделяем строку на первые 3 и последние 3 символа
start3 = ticket_number[:3]
end3 = ticket_number[-3:]

# Преобразуем каждый символ в число и считаем сумму
numbers_sum1 = sum(int(digit) for digit in start3)
numbers_sum2 = sum(int(digit) for digit in end3)

if numbers_sum1 == numbers_sum2:
    result = "Ура! Это счастливый билет!"
elif abs(numbers_sum1 - numbers_sum2) == 1:
    result = "Почти счастливый!"
else:
    result = "Обычный билет"

print(result, numbers_sum1, numbers_sum2)
```
