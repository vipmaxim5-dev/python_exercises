```python
""" Функция получения числового значения от пользователя """
def input_int(string):
    while True:
        user_input = input(string) # Получаем строку от пользователя
        try:
            # Пытаемся преобразовать в число
            integer = int(user_input) 
            return integer # Выходим из цикла и возвращаем число
        except ValueError:
            # Если возникла ошибка (ввели буквы, знаки, пробелы), пишем предупреждение
            print('Неверно, введите еще раз')
            
""" Функция получения True или False """
def input_bool(string):
    while True:
        user_input = input(string).strip().lower()
        if user_input in ("yes", "y", "да", "д"):
            return True
        elif user_input in ("no", "n", "нет", "н"):
            return False
        else:
            # Если введено что-то другое, цикл продолжится
            print("Неверно! Введите еще раз.")

temperature = input_int("Введите темпиратуру воздуха в градусах цельсия: ") # Температура в градусах
is_raining = input_bool("Идет ли дождь? (yes/y или да/д или no/n или нет/н): ") # Идет/Не идет дождь
is_windy = input_bool("Есть ли ветер? (yes/y или да/д или no/n или нет/н): ") # Ветрено/Не ветрено
have_umbrella = input_bool("У вас есть зант с собой? (yes/y или да/д или no/n или нет/н): ") # Есить ли зонт

if temperature < 0: # Если температура < 0: "Очень холодно, одевайтесь теплее!"
    result = "Очень холодно, одевайтесь теплее!"
elif 0 <= temperature <= 15 and is_raining and is_windy: # Если 0-15 и идет дождь и ветер: "Сыро и ветрено, лучше остаться дома"
    result = "Сыро и ветрено, лучше остаться дома"
elif 15 <= temperature <= 25 and not is_raining: # Если 15-25 и не идет дождь: "Отличная погода для прогулки!"
    result = "Отличная погода для прогулки!"
elif temperature > 25 and not is_raining: # Если > 25 и не идет дождь: "Жарко, не забудьте головной убор"
    result = "Жарко, не забудьте головной убор"
elif is_raining and have_umbrella: # Если идет дождь и есть зонт: "Можно идти, зонт спасет"
    result = "Можно идти, зонт спасет"
elif is_raining and not have_umbrella: # Если идет дождь и нет зонта: "Лучше переждать дождь дома"
    result = "Лучше переждать дождь дома"

print(result)
```
