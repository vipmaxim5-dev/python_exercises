```python
""" Функция получения числового значения от пользователя """
def input_int(string):
    while True:
        user_input = input(string) # Получаем строку от пользователя
        
        # Проверяем, состоит ли строка только из цифр
        if user_input.isdigit():
            integer = int(user_input) # Переводим в число
            return integer # Выходим из цикла и возвращаем число
        else:
            print('Неверно, введите еще раз')

""" Функция расчета времени """
def get_time(time_sec):
    time_h, time_sec = divmod(time_sec, 3600) # Получаем целое чисто при делении в time_h и его остаток в time_sec
    time_min, time_sec = divmod(time_sec, 60) # Получаем целое чисто при делении в time_min и его остаток в time_sec
    return time_h, time_min, time_sec # Возвращаем поделенные секунды на часы, минуты, секунды

total_seconds = input_int('Введите количество секунд (например, 3665):')
total_time = get_time(total_seconds)
print(f'Это составляет {total_time[0]} час(a/ов), {total_time[1]} минут(ы), {total_time[2]} секунд(ы)')
```
