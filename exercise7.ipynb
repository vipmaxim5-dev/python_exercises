```python
""" Функция получения числового значения от пользователя """
def input_int(string):
    error = 'Неверно, введите еще раз'
    while True:
        user_input = input(string) # Получаем строку от пользователя
        
        # Проверяем, состоит ли строка только из цифр
        if user_input.isdigit():
            integer = int(user_input) # Переводим в число
            if (0 <= integer <= 100):
                return integer # Выходим из цикла и возвращаем число
        print(error)
        
score = input_int('Введи количество баллов (от 0 до 100): ')

if 90 <= score <= 100:
    result = "Отлично (5)\nПоздравляю с отличным результатом!"
elif 75 <= score <= 89:
    result = "Хорошо (4)"
elif 50 <= score <= 74:
    result = "Удовлетворительно (3)"
else:
    result = "Неудовлетворительно (2)\nНужно подтянуть знания!"

print(f'{result}')
```
