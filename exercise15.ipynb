```python
import math

point_a = (3, 5)
point_b = (7, 2)
point_c = (1, 9)

x1, y1 = point_a
print(f'Распакованные координаты точки A: x1 = {x1}, y1 = {y1}')

x2, y2 = point_b
distance_ab = math.sqrt((x2 - x1)**2 + (y2 - y1)**2)
print(f'Расстояние между точками A и B: {distance_ab:.2f}')

dist_a = math.sqrt(point_a[0]**2 + point_a[1]**2)
dist_b = math.sqrt(point_b[0]**2 + point_b[1]**2)
dist_c = math.sqrt(point_c[0]**2 + point_c[1]**2)

distances = {point_a: dist_a, point_b: dist_b, point_c: dist_c}
furthest_point = max(distances, key=distances.get)

print(f'Самая удаленная точка: {furthest_point}')

try:
    point_a[0] = 10
except TypeError as e:
    print(f'Ошибка при изменении: {e}')

# ОБЪЯСНЕНИЕ: Кортеж (tuple) является неизменяемым (immutable) типом данных в Python. 
# В отличие от списков, после создания кортежа мы не можем добавлять, удалять или перезаписывать его элементы.
# Попытка изменить элемент по индексу вызывает исключение TypeError.
# Это свойство гарантирует безопасность данных.
```
