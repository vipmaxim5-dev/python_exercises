```python
""" Функция получения числового значения от пользователя """
def input_int(string):
    while True:
        user_input = input(string) # Получаем строку от пользователя
        
        # Проверяем, состоит ли строка только из цифр
        if user_input.isdigit():
            integer = int(user_input) # Переводим в число
            return integer # Выходим из цикла и возвращаем число
        else:
            print('Неверно, введите еще раз')

""" Функция получения типа животного от пользователя """
def input_animal_type():
     while True:
        print("«С» - Если хотите выбрать собаку")
        print("«К» - Если хотите выбрать кошку")
        user_input = input("Введите «С» или «К»:")

        # Используем .upper() и .strip() для упрощения проверки регистра и пробелов
        user_input = user_input.strip().upper()
        
        if user_input in ['С', 'C']: # Проверяем русскую и английскую раскладку, а также нижний и верхний регистр
            return 'собаки' # Выходим из цикла и возвращаем dog
        elif  user_input in ['К', 'K']: # Проверяем русскую и английскую раскладку, а также нижний и верхний регистр
            return 'кошки' # Выходим из цикла и возвращаем cat
        else:
            print('Неверно, введите еще раз')

""" Функция расчета лет животного """
def get_years_animal(animal_type, human_years):
    years_animal = 0 # Итоговый возраст питомца
    
    # Определяем шкалу возраста [1-й год, 2-й год, последующие]
    if animal_type == 'собаки':
        rates = [15, 9, 5]
    elif animal_type == 'кошки':
        rates = [15, 10, 4]

    """Считаем года"""
    # Первый год
    if human_years >= rates[0]:
        human_years -= rates[0]
        years_animal += 1
        
        # Второй год
        if human_years >= rates[1]:
            human_years -= rates[1]
            years_animal += 1
            
            # Считаем все последующие года
            while human_years >= rates[2]:
                human_years -= rates[2]
                years_animal += 1
    else:
        return 'меньше года'
            
    return years_animal

# Вызываем функции:
human_years = input_int("Введите возраст человека (например,5): ")
animal_type = input_animal_type()
years_animal = get_years_animal(animal_type, human_years)

# Выводим результат 
print (f'Возраст {animal_type} в человеческих годах: {years_animal}')
```
