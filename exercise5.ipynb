```python
MAX_HEIGHT_M = 2.72

""" Функция получения вещественного значения от пользователя """
def input_float(string):
  while True:
    user_input = input(string).replace(",", ".")  # Заменяем запятую на точку
    try:
      number = float(user_input)
      return number
    except ValueError:
      print("Неверно, введите еще раз")
    
weight_kg = input_float('Введите вес тела (вес в кг):') # Вес тела
height_m = input_float('Введите рост (рост в метрах):') # Рост
    
if height_m > MAX_HEIGHT_M: # Пользователь не внимательный и ввел рост в сантиметрах
    height_m = height_m / 100 # Переводим в метры
    
BMI = weight_kg / height_m**2

if BMI < 18.5:
    print("Недостаточный вес")
elif BMI >= 18.5 and BMI < 24.9:
    print("Нормальный вес")
elif BMI >= 25 and BMI <29.9:
    print("Избыточный вес")
else:
    print("Ожирение")
```
