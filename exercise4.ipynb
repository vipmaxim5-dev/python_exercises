```python
FIX_DISCOUNT = [10, 5]
FIX_QUANTITY = 5

""" Функция получения числового значения от пользователя """
def input_int(string):
    while True:
        user_input = input(string) # Получаем строку от пользователя
        
        # Проверяем, состоит ли строка только из цифр
        if user_input.isdigit():
            integer = int(user_input) # Переводим в число
            return integer # Выходим из цикла и возвращаем число
        else:
            print('Неверно, введите еще раз')
            
""" Функция получения True или False """
def input_bool(string):
    while True:
        user_input = input(string).strip().lower()
        if user_input in ("yes", "y", "да", "д"):
            return True
        elif user_input in ("no", "n", "нет", "н"):
            return False
        else:
            # Если введено что-то другое, цикл продолжится
            print("Неверно! Введите еще раз.")
            
""" Функция расчета скидки """
def get_discount(quantity, discount_card):
    discount = 0
    if quantity > FIX_QUANTITY:
        discount += FIX_DISCOUNT[0]
    if discount_card == True:
        discount += FIX_DISCOUNT[1]
    return discount
    
price = input_int('Введите цену товара (например, 50):') # Цена товара
quantity = input_int('Введите количество товара (например, 10):') # Количество товара
has_discount_card = input_bool('Есть ли у вас карта скидки? (yes/y или да/д или no/n или нет/н):') # Есть ли карта скидки

discount = get_discount(quantity, has_discount_card) # Считаем % скидки
total_price = price-price*discount/100 # считаем стоимость с учетом скидки

print(f'Стоимость товара без скидки: {price}')
print(f'Сумма скидки: {discount}%')
print(f'Итоговая стоимость: {total_price}')
```
