```python
text = input('Введите любой текст: ')
print(f'Кол-во символов: {len(text)}') # Кол-во символов
print(f'Кол-во символов без пробелов: {len(text.replace(" ", ""))}') # Кол-во символов без пробелов
words = text.split() # Разделить по словам
print(f'Самое длинное слово: { max(words, key=len)}') # Cамое длинное слово
print(f'Все пробелы теперь символ подчеркивания: {text.replace(" ", "_")}') # Пробелы на символ подчеркивания
clean_text = text.lower().replace(" ", "") # Приводим к нижнему регистру и удаляем пробелы
is_palindrome = clean_text == clean_text[::-1] # Проверяем, читается ли строка одинаково в обоих направлениях
print(f'Является ли текст палиндромом: {is_palindrome}')
print(f'Текст в обратном порядке: {text[::-1]}')
```
