```python
""" Функция получения вещественного значения от пользователя """
def input_float(string):
  while True:
    user_input = input(string).replace(",", ".")  # Заменяем запятую на точку
    try:
      number = float(user_input)
      return number
    except ValueError:
      print("Неверно, введите еще раз")

a, b, c = input_float('Введите сторону a: '), input_float('Введите сторону b: '), input_float('Введите сторону c: ')

if a < b + c and b < a + c and c < a + b:
    print("Треугольник существует!")
    if a == b == c:
        print('Равносторонний')
    elif a == b or b == c or a == c:
        print('Равнобедренный')
    else:
        print('Разносторонний')
else:
    print("Ошибка: треугольник с такими сторонами не существует.")
```
