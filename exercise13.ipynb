```python
instagram_friends = {"Анна", "Петр", "Мария", "Иван", "Елена"}
vk_friends = {"Петр", "Дмитрий", "Елена", "Ольга", "Сергей"}
telegram_friends = {"Анна", "Сергей", "Елена", "Михаил", "Ирина"}

common_friends = instagram_friends.intersection(vk_friends, telegram_friends) # Поиск друзей, которые есть во всех трех соцсетях (пересечение)
print(f'Друзья во всех трех соцсетях: {common_friends}')

unique_instagram = instagram_friends.difference(vk_friends, telegram_friends) # Поиск уникальных друзей в Instagram (которых нет в других сетях)
print(f'Уникальные друзья в Instagram: {unique_instagram}')

all_friends = instagram_friends.union(vk_friends, telegram_friends) # Поиск друзей, которые есть хотя бы в одной соцсети (объединение)
print(f'Все друзья (хотя бы в одной соцсети): {all_friends}')

has_alexey = "Алексей" in all_friends # Проверка, есть ли друг "Алексей" хотя бы в одной соцсети
print(f'Есть ли Алексей в списках?: {has_alexey}')
```
