```python
shopping_list = ["хлеб", "молоко", "яйца", "сыр", "помидоры"]
print(f'Изначальный список:\n{shopping_list}\n')

shopping_list += ["огурцы", "колбаса"]
#shopping_list.append("огурцы")
#shopping_list.append("колбаса")
print(f'Добавлено в список "огурцы" и "колбасу":\n{shopping_list}\n')

shopping_list.remove("сыр")
print(f'Удалено из списка "сыр":\n{shopping_list}\n')

shopping_list[shopping_list.index("помидоры")] = "перец"
print(f'Заменено "помидоры" на "перец":\n{shopping_list}\n')

shopping_list.sort()
print(f'Список в алфавитном порядке:\n{shopping_list}\n')

print(f'Количество элементов в списке: {len(shopping_list)}\n')

print(f'Первые 3 товара из списка:\n{shopping_list[:3]}')
```
